# Dimensionality Reduction



Dimensionality Reduction is a technique used to reduce the number of input features (columns) in a dataset while keeping the important information.

Why Dimensionality Reduction is Used

1. Reduces computational time and memory usage.

2. Removes unnecessary or redundant features.

3. Helps avoid overfitting in machine learning models.

4. Improves model performance and accuracy.

5. Makes data visualization easier (2D or 3D plots).

6. Reduces noise in the dataset.

7. Helps handle high-dimensional data efficiently.

Common Dimensionality Reduction Techniques

1. PCA (Principal Component Analysis)
2. LDA (Linear Discriminant Analysis)
3. t-SNE
4. UMAP


# PCA (Principal Component Analysis)



PCA is a dimensionality reduction technique used to reduce the number of features in a dataset while preserving the maximum important information.

How PCA Works

1. PCA converts original features into new features called Principal Components.

2. These components are created in such a way that:
   - First component captures maximum variance.
   - Second component captures second highest variance.
   - And so on.

3. Principal components are uncorrelated with each other.

Why PCA is Used

1. Reduces dimensionality of data.

2. Speeds up machine learning algorithms.

3. Reduces overfitting.

4. Removes multicollinearity.

5. Helps in data visualization.

6. Improves model efficiency.

Advantages of PCA

1. Faster training time.
2. Reduces storage space.
3. Removes noise from data.
4. Works well with high-dimensional datasets.

Disadvantages of PCA

1. Reduced interpretability of features.
2. Some information loss may occur.
3. Difficult to explain principal components.
4. Sensitive to feature scaling.

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
X, y = make_blobs(
    n_samples=300,
    n_features=4,
    centers=3,
    random_state=42
)
```

```python
df = pd.DataFrame(
    X,
    columns=['Feature_1', 'Feature_2', 'Feature_3', 'Feature_4']
)
```

```python
df.head()
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(df)
```

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

# Convert PCA result into dataframe
pca_df = pd.DataFrame(
    X_pca,
    columns=['PC1', 'PC2']
)
```

```python
pca_df.head()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x=pca_df['PC1'],
    y=pca_df['PC2'],
    hue=y,
    palette='Set1',
    s=100
)

plt.title("PCA Visualization")

plt.show()
```
